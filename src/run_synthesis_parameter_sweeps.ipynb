```python
def run_sweeps():
    sweeps = {
        # From largest host star mass to lowest.
        'm1p2_atm': sweep_earth_m1p2_atm,
        'm1p2_water': sweep_earth_m1p2_water,
        'm1p1_atm': sweep_earth_m1p1_atm,
        'm1p1_water': sweep_earth_m1p1_water,
        'sun_atm': sweep_earth_sun_atm,
        'sun_water': sweep_earth_sun_water,
        'm0p9_atm': sweep_earth_m0p9_atm,
        'm0p9_water': sweep_earth_m0p9_water,
        'kv_atm': sweep_earth_kv_atm,
        'kv_water': sweep_earth_kv_water,
        'm0p8_atm': sweep_earth_m0p8_atm,
        'm0p8_water': sweep_earth_m0p8_water,
        'm0p7_atm': sweep_earth_m0p7_atm,
        'm0p7_water': sweep_earth_m0p7_water,
        'm0p6_atm': sweep_earth_m0p6_atm,
        'm0p6_water': sweep_earth_m0p6_water,
        'm0p5_atm': sweep_earth_m0p5_atm,
        'm0p5_water': sweep_earth_m0p5_water,
        'm0p4_atm': sweep_earth_m0p4_atm,
        'm0p4_water': sweep_earth_m0p4_water,
        'm0p3_atm': sweep_earth_m0p3_atm,
        'm0p3_water': sweep_earth_m0p3_water,
        'm0p2_atm': sweep_earth_m0p2_atm,
        'm0p2_water': sweep_earth_m0p2_water,
        'm0p1_atm': sweep_earth_m0p1_atm,
        'm0p1_water': sweep_earth_m0p1_water,
        'trappist_atm': sweep_earth_trappist_atm,
        'trappist_water': sweep_earth_trappist_water,
    }

    for (sweep_dir, sweep_parameters) in sweeps.items():
        if sweep_dir != "kv_atm":
            continue

        Parameter_Sweep(**sweep_parameters).generate_input_files(path("data", "parameter_sweeps", "synthesis", sweep_dir))
        run_parameter_sweep()

        break

run_sweeps()
```
